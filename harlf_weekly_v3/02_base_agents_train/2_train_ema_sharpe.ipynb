## 📊 Understanding EMA Sharpe

The EMA Sharpe reward uses:
- **Exponential Moving Average** to track mean and variance of returns
- **Rolling window** to control memory (default: 12 weeks)
- **Sharpe-like calculation**: `reward = return / volatility * sqrt(52)`

### Formula (from rewards.py):
```python
delta = port_return - self._ema_mean
self._ema_mean += self.alpha * delta
self._ema_var = (1.0 - self.alpha) * (self._ema_var + self.alpha * delta * delta)

vol = sqrt(self._ema_var)
reward = port_return / vol * sqrt(52)
```

### When to Use:
- ✅ Good for stable, risk-adjusted returns
- ✅ Balances return and volatility automatically
- ✅ Fast to train
- ❌ May be too conservative in some markets
