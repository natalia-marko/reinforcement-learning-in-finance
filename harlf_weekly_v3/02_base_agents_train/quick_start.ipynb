# Quick Start Guide

## 🚀 Super Simple - Just 3 Steps!

This notebook shows you how easy it is to train your agents.

## Step 1: Import

Just one import!

```python
from train import train
```

## Step 2: Train

Pick your:
- Agent type: `'technical'` or `'sentiment'`
- Algorithm: `'PPO'`, `'SAC'`, or `'A2C'`
- Reward: `'ema_sharpe'`, `'differential_sharpe'`, `'multi_objective'`, `'simple_return'`, or `'simple_sharpe'`

```python
# Train technical agent with EMA Sharpe reward using PPO
result = train(
    agent_type='technical',
    algorithm='PPO',
    reward_type='ema_sharpe',
    verbose=True
)
```

## Step 3: Check Results

That's it! Results are printed automatically.

```python
# You can also access the results dict
print(f"\nTest Sharpe: {result['test_sharpe']:.3f}")
print(f"Model saved at: {result['model_path']}")
```

## 🎯 More Examples

Try different combinations:

```python
# Sentiment agent with differential Sharpe
result2 = train('sentiment', 'SAC', 'differential_sharpe')

# Technical agent with multi-objective
result3 = train('technical', 'A2C', 'multi_objective')

# Simple baseline
result4 = train('technical', 'PPO', 'simple_return')
```
